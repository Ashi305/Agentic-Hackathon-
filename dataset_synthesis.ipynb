```python
!pip install faker pandas

import pandas as pd
import random
from faker import Faker
from datetime import datetime, timedelta

fake = Faker()

# Configuration
NUM_RECORDS = 5000 # Increased to show scale
CRITICAL_RATIO = 0.15

locations = [
    "Warehouse A", "Warehouse B", "Loading Dock North", "Loading Dock South",
    "Production Line 1", "Production Line 2", "Production Line 3", "Chemical Storage",
    "Maintenance Shop", "Server Room", "Main Office", "Breakroom", "Hallway C",
    "Boiler Room", "Rooftop HVAC Unit", "Cold Storage", "Assembly Station Alpha",
    "R&D Lab", "Paint Shop", "Welding Bay", "Visitor Parking Lot",
    "Hazardous Waste Bay", "Security Desk", "Mezzanine Level", "Freight Elevator"
]

departments = [
    "Logistics", "Manufacturing", "Facilities", "IT", "Operations",
    "Admin", "Quality Assurance", "Health & Safety (EHS)", "R&D",
    "Engineering", "Sanitation", "Security", "Human Resources", "Fleet Management"
]

contexts = [
    "During my routine walkthrough, ",
    "While operating the equipment, ",
    "At the start of the morning shift, ",
    "I was walking back from the breakroom and ",
    "While doing some cleanup, ",
    "Just before heading home, ",
    "During the mid-day inspection, ",
    "While unloading the delivery truck, ",
    "Responding to a strange noise, ",
    "During the shift handover, ",
    "While calibrating the sensors, ",
    "Doing a random spot check, ",
    "Walking the perimeter, ",
    "While escorting a contractor, ",
    "Just after lunch, ",
    "During the late night shift, ",
    "Following up on a previous ticket, ",
    "While conducting the monthly audit, ",
    "" # Raw start
]

observations = {
    "Critical": {
        "Electrical": [
            "I noticed an exposed frayed wire near the main water line.",
            "there was a distinct burning smell and buzzing coming from the breaker panel.",
            "the grounding cable on the main press was completely disconnected.",
            "sparks were actively shooting out of the 220V wall receptacle.",
            "water from a leaky roof was dripping directly onto a live, uncovered junction box.",
            "someone had daisy-chained five power strips together to run a heavy-duty industrial compressor.",
            "the main emergency electrical disconnect switch was broken off in the 'ON' position."
        ],
        "Slip/Trip/Fall": [
            "there was a massive hydraulic oil puddle that hadn't been cleaned or marked.",
            "someone left a pallet of heavy materials wedged right in front of the emergency fire exit.",
            "the safety guardrail on the second-story mezzanine had been completely removed.",
            "a deep sinkhole/crack is forming in the main forklift traffic lane.",
            "the scaffolding for the overhead repair looked visibly bowed and structurally compromised."
        ],
        "Chemical/Hazmat": [
            "a secondary containment drum was overflowing with an unknown, pungent solvent.",
            "I saw new contractors mixing volatile cleaning agents without wearing any respirators.",
            "there was a steady, corrosive drip from the overhead sulfuric acid pipe.",
            "three unlabelled, visibly bulging steel drums were left sitting out in the hot sun.",
            "a very strong ammonia smell was making everyone's eyes water in the sector.",
            "a forklift clipped the propane tank cage and there is a loud hissing sound."
        ],
        "Equipment/LOTO": [
            "the laser safety guard on the metal lathe had been intentionally bypassed with black tape.",
            "the overhead crane hoist chain was heavily corroded, bent, and missing a safety latch.",
            "the forklift brakes failed completely during a stop, sliding 10 feet.",
            "an operator was reaching in to clean the mixing blades while the machine was still plugged in.",
            "the pressure release valve on the main boiler was completely rusted shut.",
            "a driver was speeding a forklift blindly around a blind corner with the forks raised high."
        ],
        "Fire/Explosion": [
            "a tall stack of cardboard boxes was pushed directly against an active space heater.",
            "oily shop rags were actively smoldering inside a regular plastic trash can.",
            "the only fire extinguisher for the welding bay was depressurized and missing its pin.",
            "the emergency fire doors at the end of the hall were chained and padlocked shut.",
            "someone was grinding metal directly next to the open paint-thinner vats."
        ]
    },
    "Ordinary": {
        "Housekeeping": [
            "the main trash bin was overflowing with paper towels.",
            "someone left their empty coffee cups on the shared workstation.",
            "there was a visible layer of dust buildup on the ceiling ventilation grate.",
            "a wet mop was left leaning against the wall instead of being hung up.",
            "there were a few scattered zip ties left on the floor near the racks.",
            "muddy footprints were tracked all through the main visitor lobby.",
            "the blue recycling bin is full and needs to be emptied.",
            "the whiteboard in the meeting room wasn't erased after the morning huddle.",
            "the paper shredder bin is slightly overflowing onto the carpet."
        ],
        "Minor Maintenance": [
            "one of the fluorescent overhead lights is flickering annoyingly.",
            "the door hinge to the breakroom is squeaking quite loudly.",
            "the yellow paint on the pedestrian walkway lines is starting to fade.",
            "the soap dispenser in the men's restroom is empty again.",
            "the doorknob to the supply closet feels a bit loose.",
            "the bathroom sink faucet has a slow, steady drip.",
            "the paint is peeling off the handrail near the back stairs.",
            "the AC in the admin wing seems to be running a bit warm today.",
            "the wall clock in the cafeteria has a dead battery."
        ],
        "Minor Slip/Trip": [
            "a few drops of water were on the floor near the breakroom sink.",
            "an extension cord was temporarily taped across the floor for the vacuum.",
            "a small piece of packing cardboard was left on the floor near the printers.",
            "a stray paperclip was on the ground in the hallway.",
            "the welcome floor mat in the hallway was slightly bunched up at the corner."
        ],
        "Behavioral": [
            "someone forgot to push their ergonomic chair back in under the desk.",
            "a worker was wearing their safety glasses on their forehead while filling out paperwork.",
            "people were talking quite loudly near the designated quiet zone.",
            "an employee was walking while texting in the admin corridor.",
            "someone was wearing earbuds in both ears while at their desk.",
            "someone left their lunch in the microwave for too long and it smells."
        ]
    }
}

consequences = {
    "Critical": [
        " If someone hadn't noticed, this could have caused a fatal incident.",
        " Someone could have easily lost a limb or worse.",
        " This is a massive fire hazard waiting to happen.",
        " This is a serious precursor to a catastrophic mechanical failure.",
        " It's a miracle no one was standing underneath when it happened.",
        " This could lead to a severe chemical burn or respiratory damage.",
        " We could be facing a massive environmental spill fine.",
        " This is an immediate OSHA shutdown violation.",
        " Could result in permanent eye damage or blindness."
    ],
    "Ordinary": [
        " Not a huge deal but should be fixed when there is time.",
        " Just wanted to note it for the next routine maintenance cycle.",
        " Looks a bit untidy and unprofessional.",
        " Someone might stub their toe if they aren't looking.",
        " It's just a minor inconvenience.",
        " Purely a cosmetic issue, but worth logging.",
        " Annoying squeak, but harmless.",
        "",
        ""
    ]
}

actions = {
    "Critical": [
        " I immediately locked out the machine (LOTO) and called the shift supervisor.",
        " Put up red caution tape and logged a high-priority, urgent maintenance ticket.",
        " Evacuated the immediate area, hit the emergency stop, and called security.",
        " I confronted the operator, halted work, and reminded them of the life-saving protocols.",
        " Immediately shut off the main gas valve and opened the bay doors for ventilation.",
        " Hit the fire alarm and grabbed the nearest extinguisher just in case.",
        " Stood guard to prevent anyone from entering the zone while waiting for Hazmat."
    ],
    "Ordinary": [
        " I just picked it up myself and threw it away.",
        " Put in a low-priority work order in the system.",
        " Wiped it up with a paper towel on my way past.",
        " Reminded them politely to fix it next time.",
        " No action taken, just reporting it for the record.",
        " Left a sticky note on it for the janitorial staff.",
        " Sent a quick email to facilities about it.",
        " Flattened the mat out with my foot and kept walking."
    ]
}

def inject_noise(text):
    if random.random() < 0.2: text = text.lower()
    if random.random() < 0.15: text = text.replace(".", "")
    if random.random() < 0.1: text = text.replace(",", "")
    if random.random() < 0.05:
        typos = {"machine": "machin", "the": "teh", "safety": "safty", "maintenance": "maintnance", "hazard": "hazrd"}
        for k, v in typos.items(): text = text.replace(k, v)
    return text

def generate_report():
    severity = "Critical" if random.random() < CRITICAL_RATIO else "Ordinary"
    category = random.choice(list(observations[severity].keys()))

    context = random.choice(contexts)
    observation = random.choice(observations[severity][category])
    consequence = random.choice(consequences[severity])
    action = random.choice(actions[severity])

    # Randomly omit consequences or actions to simulate incomplete reporting
    if random.random() < 0.2: consequence = ""
    if random.random() < 0.1: action = ""

    raw_text = f"{context}{observation}{consequence}{action}"
    narrative = inject_noise(raw_text)

    days_ago = random.randint(0, 1000)
    report_date = datetime.now() - timedelta(days=days_ago)

    return {
        "Report_ID": fake.uuid4()[:10].upper(),
        "Date": report_date.strftime("%Y-%m-%d"),
        "Location": random.choice(locations),
        "Department": random.choice(departments),
        "Category_Tag": category if random.random() > 0.08 else "Uncategorized",
        "Ground_Truth_Severity": severity,
        "Report_Text": narrative.strip()
    }

# Generate dataset
data = [generate_report() for _ in range(NUM_RECORDS)]
df = pd.DataFrame(data)

# Save to CSV
file_path = "extensive_safety_reports.csv"
df.to_csv(file_path, index=False)

print(f"✅ Generated {NUM_RECORDS} complex reports.")
print(df['Ground_Truth_Severity'].value_counts())
```
